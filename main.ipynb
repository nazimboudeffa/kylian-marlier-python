```python
import pandas as pd

# Chargement du fichier CSV (remplace 'ton_fichier.csv' par le nom réel si besoin)
df = pd.read_csv('202505.csv')

# Affichage des premières lignes
print(df.head())
```

```python
# Infos de base
print(df.info())

# Statistiques descriptives
print(df.describe())
```

```python
# Moyenne du signal
print("Moyenne du signal :", df['signal'].mean())

# Valeur max et min
print("Signal maximum :", df['signal'].max())
print("Signal minimum :", df['signal'].min())

# Répartition positive vs négative
print("Signaux positifs :", (df['signal'] > 0).sum())
print("Signaux négatifs :", (df['signal'] < 0).sum())
```

```python
# Paires avec les signaux les plus forts (positifs ou négatifs)
df_sorted = df.sort_values(by='signal', ascending=False)
print(df_sorted)
```

```python
# Nombre de paires distinctes
nb_pairs_uniques = df['pair'].nunique()
print("Nombre de paires distinctes :", nb_pairs_uniques)
```

```python
# Fréquence des paires tradées
frequences = df['pair'].value_counts()
print(frequences)
```

```python
import matplotlib.pyplot as plt

frequences.plot(kind='bar', figsize=(10, 5))
plt.title('Fréquence des paires tradées')
plt.xlabel('Paire de devises')
plt.ylabel('Nombre de signaux')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Total des pips
total_pips = df['signal'].sum()
print("Total de pips :", total_pips)
```
